```python
#Question 01 Student result management
students = []

def result(marks):
    total = sum(marks)
    per = total / 3

    if per >= 80:
        grade = "A"
    elif per >= 70:
        grade = "B"
    elif per >= 60:
        grade = "C"
    elif per >= 50:
        grade = "D"
    else:
        grade = "F"

    return total, per, grade


for i in range(5):
    name = input("Name: ")
    roll = input("Roll No: ")

    marks = [
        float(input("Python: ")),
        float(input("AI: ")),
        float(input("Math: "))
    ]

    total, per, grade = result(marks)

    students.append({
        "Name": name, "Roll": roll,
        "Marks": marks, "Total": total,
        "Percentage": per, "Grade": grade
    })

for s in students:
    print(s)

top = max(students, key=lambda x: x["Percentage"])
print("Highest:", top["Name"], top["Percentage"])

with open("results.txt", "w") as f:
    for s in students:
        f.write(str(s) + "\n")
```

```python
#Question 2 sentence analytics
def analyze(sentence):
    words = sentence.split()

    print("Words:", len(words))
    print("Characters:", len(sentence))
    print("Longest:", max(words, key=len))
    print("Shortest:", min(words, key=len))

    vowels = 0
    for ch in sentence.lower():
        if ch in "aeiou":
            vowels += 1

    print("Vowels:", vowels)
    print("Unique:", set(words))
    print("Alphabetical:", sorted(set(words)))

    with open("word_analysis.txt", "w") as f:
        f.write("Words: " + str(len(words)) + "\n")
        f.write("Characters: " + str(len(sentence)) + "\n")
        f.write("Longest: " + max(words, key=len) + "\n")
        f.write("Shortest: " + min(words, key=len) + "\n")
        f.write("Vowels: " + str(vowels) + "\n")
        f.write("Unique: " + str(set(words)) + "\n")
        f.write("Alphabetical: " + str(sorted(set(words))))

sentence = input("Enter sentence: ")
analyze(sentence)
```

```python
#Question 3 shopping program
products = {}

def bill():
    total = 0

    for p in products:
        total += products[p]["price"] * products[p]["qty"]

    return total


for i in range(5):
    name = input("Product: ")
    price = float(input("Price: "))
    qty = int(input("Quantity: "))

    products[name] = {"price": price, "qty": qty}

total = bill()

discount = total * 0.10 if total > 10000 else 0
final = total - discount

print("Bill:", total)
print("Discount:", discount)
print("Final Bill:", final)

expensive = max(products, key=lambda x: products[x]["price"])
print("Most expensive:", expensive)

with open("bill.txt", "w") as f:
    f.write("Bill: " + str(total))
    f.write("\nDiscount: " + str(discount))
    f.write("\nFinal Bill: " + str(final))
    f.write("\nMost expensive: " + expensive)
```

```python
#Question 4 calculator
def add(a,b): return a+b
def sub(a,b): return a-b
def mul(a,b): return a*b
def div(a,b): return a/b
def mod(a,b): return a%b

while True:
    a = float(input("First number: "))
    b = float(input("Second number: "))
    op = input("Operation (+,-,*,/, %): ")

    if op == "+": print(add(a,b))
    elif op == "-": print(sub(a,b))
    elif op == "*": print(mul(a,b))
    elif op == "/": print(div(a,b))
    elif op == "%": print(mod(a,b))
    else: print("Invalid operation")

    again = input("Again? yes/no: ")
    if again.lower() != "yes":
        break
```

```python
#Question 5 resturantant program
menu = {
    "Burger": 500,
    "Pizza": 800,
    "Fries": 250,
    "Drink": 150
}

orders = {}

def bill():
    total = 0
    for item in orders:
        total += menu[item] * orders[item]
    return total


while True:
    print("\n1.Burger  2.Pizza  3.Fries  4.Drink  5.Exit")
    choice = input("Choose: ")

    if choice == "1": item = "Burger"
    elif choice == "2": item = "Pizza"
    elif choice == "3": item = "Fries"
    elif choice == "4": item = "Drink"
    elif choice == "5": break
    else:
        print("Invalid")
        continue

    qty = int(input("Quantity: "))
    orders[item] = orders.get(item, 0) + qty

total = bill()

print("Final Bill:", total)

with open("orders.txt", "w") as f:
    f.write("Orders: " + str(orders))
    f.write("\nFinal Bill: " + str(total))
```
